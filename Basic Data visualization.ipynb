1- Univariate - Quality of one feature (Countplot, piechart, histogram)
2. Bivariate - Quality/relationship between two feature (Bar plot and scatter plot)
3. Multivariate - Relationship between more than two features (Pair plot, Heatmap, line-plot(data time series analysis))

### About mtcars dataset

The mtcars dataset is a classic, built-in data frame from the 1974 Motor Trend US magazine. 
It tracks 11 distinct design and performance features for 32 different automobiles from the 1973–74 model years.

Dataset Variables
mpg: Miles per gallon (fuel efficiency)
cyl: Number of cylinders (4, 6, or 8)
disp: Engine displacement in cubic inches
hp: Gross horsepowerdrat: Rear axle ratio
wt: Weight measured in 1,000 lb
sqsec: Time to complete a 1/4 mile
vs: Engine shape/type (0 = V-shaped, 1 = straight/inline)
am: Transmission type (0 = automatic, 1 = manual)
gear: Number of forward gears
carb: Number of carburetors

### Discrete vs continous?

1.Discrete- countable (Fixed distinct points)
2.Continous - Measurable (Infinite possibilities with a range) - 
continous data can be splitted in to decimals or anyvalues and will have units mostly


### What type of feature or data is needed to display histogram?

Only takes on continous feature or data. it would give wrong visualization if you give discrete data

### For the above hist i need to delete the array, need a title MPG distribution, xasis named as MPEG and Y axis named as frequency?

```python
plt.hist(mtcars_data["mpg"])
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.title("MPG distibution")
plt.show() #removes the array involved and only shows the figure
```
