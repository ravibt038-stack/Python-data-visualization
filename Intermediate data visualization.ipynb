### pie chart? What is very important for pie chart?

    You need to get the count of a single discrete data that needs to be displayed in the chart, without count it wont appear
    Pie chart is to understand the distibution of the value and the count of a discrete value in a percentage manner

```python
Cylinder=mtcars_data["cyl"].value_counts()
plt.pie(x=Cylinder,labels=Cylinder.index)
```

```python
plt.pie(x=Cylinder,labels=["8 cylinder","4 cylinder","6 cylinder"])
```

```python
plt.pie(x=Cylinder,labels=["8 cylinder","4 cylinder","6 cylinder"],explode=[0.01,0.01,0.01])
```

### I need to increase the size of the figure how is it done?

```python

plt.pie(x=Cylinder,labels=["8 cylinder","4 cylinder","6 cylinder"],explode=[0.01,0.01,0.01], autopct="%1.1f%%")
plt.figure(figsize=[7,7])
```

```python


plt.figure(figsize=[7,7])
plt.pie(x=Cylinder,labels=["8 cylinder","4 cylinder","6 cylinder"],explode=[0.01,0.01,0.01], autopct="%1.1f%%")
plt.show()
```

### Pearson correlation coefficient

Understanding the Score 
(r)+1: Both variables go up together in a straight line.
(r) -1: One variable goes up as the other goes down in a straight line.
R=0: No straight-line pattern exists between the two sets of data.

### What are the ranges for the correlation how to interpret the above value?

NOTES: this appplies to both negative and positive correlation parameters
. 0 to 0.20 = Very Weak Correlation
. 0.21 to 0.40 = Weak Correlation
. 0.41 to 0.60 = Normal/Standard Correlation
. 0.61 to 0.80 = Strong Cotvelation
. 0.81 to 1 = Very Strong Correlation

### How to get penguin data from seaborn and do histogram with continous feature as flipper_lenghth_mm with kde,bins and color?

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load example dataset
penguins = sns.load_dataset("penguins")

# Plot histogram with KDE
sns.histplot(data=penguins, x="flipper_length_mm", kde=True, bins=20, color="skyblue")
plt.title("Flipper Length Distribution with KDE")
plt.show()
```

### How to get penguin data from seaborn and do histogram with continous feature and a discrete feature 
### as flipper_lenghth_mm with kde,bins and color?

```python
sns.histplot(data=penguins, x="flipper_length_mm", hue="species", multiple="stack")
```

```python
sns.histplot(data=penguins, x="flipper_length_mm", hue="species", multiple="layer")
```
